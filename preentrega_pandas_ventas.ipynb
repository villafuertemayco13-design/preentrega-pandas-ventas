```python

# Creamos los datos de las ventas
ventas = pd.DataFrame({
    "id_producto": [101, 102, 103, 101, 104, 102, 105],
    "cantidad": [2, 1, 3, 1, 2, 4, 1],
    "fecha": [
        "2026-10-01", "2026-10-02", "2026-10-03",
        "2026-10-04", "2026-10-05", "2026-10-06",
        "2026-10-07"
    ]
})

# Creamos el catálogo de productos
productos = pd.DataFrame({
    "id_producto": [101, 102, 103, 104, 105],
    "nombre_producto": [
        "Perfume", "Crema facial", "Shampoo",
        "Desodorante", "Serum facial"
    ],
    "precio_unitario": [45000, 18000, 9500, 6500, 32000]
})

# Guardamos los archivos en distintos formatos
ventas.to_csv("ventas.csv", index=False)
productos.to_excel("productos.xlsx", index=False)

print("¡Archivos creados correctamente!")
```

```python

import pandas as pd

# 1. Cargar los archivos
ventas = pd.read_csv("ventas.csv")
productos = pd.read_excel("productos.xlsx")

# 2. Corregir los tipos de datos
ventas["fecha"] = pd.to_datetime(ventas["fecha"], errors="coerce")
ventas["id_producto"] = pd.to_numeric(ventas["id_producto"], errors="coerce")
productos["id_producto"] = pd.to_numeric(productos["id_producto"], errors="coerce")
ventas["cantidad"] = pd.to_numeric(ventas["cantidad"], errors="coerce")
productos["precio_unitario"] = pd.to_numeric(productos["precio_unitario"], errors="coerce")

# 3. Eliminar registros con datos críticos faltantes
ventas = ventas.dropna(subset=["id_producto", "cantidad", "fecha"])
productos = productos.dropna(
    subset=["id_producto", "nombre_producto", "precio_unitario"]
)

# 4. Unir ventas con el catálogo de productos
dataset = pd.merge(
    ventas,
    productos,
    on="id_producto",
    how="inner",
    validate="many_to_one"
)

# 5. Calcular el total de cada venta
dataset["total_venta"] = (
    dataset["cantidad"] * dataset["precio_unitario"]
)

# 6. Exportar a formato Parquet
dataset.to_parquet("ventas_normalizadas.parquet", index=False)

# 7. Mostrar el resultado
print("Dataset final:")
display(dataset)

print("Valores nulos:")
print(dataset.isnull().sum())

print("Total facturado:", dataset["total_venta"].sum())
print("Archivo Parquet generado correctamente.")
```
